```python
"""
Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario
con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.
"""
# He asumido que las mayusculas y minusculas cuentan como letras separadas

def frecuencia_letras(cadena):
    resultado = {}
    for letra in cadena:
        if not letra.isspace():
            resultado[letra] = resultado.get(letra, 0) + 1
    return resultado

# cadena = "hola Hola"
# print(frecuencia_letras(cadena))

```

```python
"""
Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map().
"""
# He usado lambda en vez de crear una funcion aparte porque es más cómodo

def doblar_lista(numeros):
    return list(map(lambda x: x * 2, numeros))

# numeros = [6,0,4,1,1,6,1,5,0]
# print(doblar_lista(numeros))
```

```python
"""
Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros. 
La función debe devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.
"""
# He asumido que es case-sensitive y que en el caso de que el objetivo sea una cadena vacia dé error

def in_lista(palabras, objetivo):
    if not palabras:
        return []
    return list(filter(lambda x : objetivo in x, palabras))

# palabras = ["cafe", "colacao", "cereal", "canela", "galleta"]
# print(in_lista(palabras, "ca"))
```
